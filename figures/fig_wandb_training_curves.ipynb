**WandB training curves without manual CSV download**

Run `wandb login` once before using this notebook.

```python
from pathlib import Path
import os
import sys

FIGURES_DIR = Path.cwd()
if FIGURES_DIR.name != "figures" and (FIGURES_DIR / "figures").exists():
    FIGURES_DIR = FIGURES_DIR / "figures"
os.chdir(FIGURES_DIR)
sys.path.insert(0, str(FIGURES_DIR))

import importlib
import wandb_training_curves
importlib.reload(wandb_training_curves)

from wandb_training_curves import (
    DEFAULT_ENTITY,
    build_plot_config,
    fetch_wandb_training_curves,
    plot,
    print_performance_table,
)

ENTITY = DEFAULT_ENTITY
```

**Stag Hunt**

```python
SH_DIR = Path("wandb_cache/SH")

fetch_wandb_training_curves(
    preset="sh",
    entity=ENTITY,
    output_dir=SH_DIR,
    state="finished",
)

sh_config = build_plot_config("sh", SH_DIR)
plot(sh_config, figsize=(8, 6), window_size=20, xlim=(0, 3e6))
print_performance_table(sh_config, window_size=10, last_n=200)
```

**Stag Hunt with Observation Noise**

This cell collects only finished StagHunt runs whose run name ends with `_noise0.1`.

```python
SH_NOISE_DIR = Path("wandb_cache/SH_noise0.1")

fetch_wandb_training_curves(
    preset="sh-noise",
    entity=ENTITY,
    output_dir=SH_NOISE_DIR,
    state="finished",
)

sh_noise_config = build_plot_config("sh-noise", SH_NOISE_DIR)
plot(sh_noise_config, figsize=(8, 6), window_size=20, xlim=(0, 3e6))
print_performance_table(sh_noise_config, window_size=10, last_n=200)
```

**SMAX**

Edit `SMAX_FILTER` to choose one simulation, for example `['2s3z', 'tar_w=4']` or `['smacv2_5_units', 'tar_w=3']`. Use an empty list to include every finished SMAX run in the project.

```python
SMAX_FILTER = ["3m", "tar_w=2"]
filter_name = "all" if not SMAX_FILTER else "_".join(s.replace("=", "") for s in SMAX_FILTER)
SMAX_DIR = Path("wandb_cache") / f"SMAX_{filter_name}"

fetch_wandb_training_curves(
    preset="smax",
    entity=ENTITY,
    output_dir=SMAX_DIR,
    state="finished",
    contains=SMAX_FILTER,
)

smax_config = build_plot_config("smax", SMAX_DIR)
plot(smax_config, figsize=(8, 6), window_size=20, xlim=(0, 1e7))
print_performance_table(smax_config, window_size=10, last_n=200)
```
