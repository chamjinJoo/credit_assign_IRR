```python
def plot(plot_config, window_size=50, figsize=(8,6), ylim=(0,1)):
    """
    plot_config: list of dicts, each with:
      - path       : str, CSV file path
      - label      : str, legend label
      - keyword    : str, column keyword to match (e.g. 'success_rate', 'PCI_rippo')
      - color      : str, matplotlib color
      - linestyle  : str, optional (default '-')
      - linewidth  : float, optional (default 3)  # Increased for better visibility
      - alpha      : float, optional fill_between alpha (default 0.08)  # Reduced for less overlap
      - exclude    : str, optional keyword to exclude from column matching
    """
    plt.close('all')
    plt.rc('font', size=14)
    plt.rc('axes', labelsize=14)
    plt.rc('xtick', labelsize=14)
    plt.rc('ytick', labelsize=14)
    plt.rc('legend', fontsize=14)

    fig, ax = plt.subplots(figsize=figsize)

    for entry in plot_config:
        file_path = entry['path']
        label     = entry['label']
        keyword   = entry.get('keyword', 'success_rate')
        color     = entry['color']
        lsty      = entry.get('linestyle', '-')
        lw        = entry.get('linewidth', 3)  # Increased default linewidth
        alpha     = entry.get('alpha', 0.08)  # Reduced default alpha for less overlap
        exclude   = entry.get('exclude', None)

        try:
            df = pd.read_csv(file_path)
        except FileNotFoundError:
            print(f'[WARN] File not found: {file_path}')
            continue

        # column matching
        clean_kw = keyword.lower().replace('_', '').replace('-', '')
        cols = []
        for col in df.columns:
            col_lower = col.lower()
            if 'success_rate' not in col_lower: continue
            clean_col = col_lower.replace('_', '').replace('-', '')
            if clean_kw not in clean_col: continue
            if 'MIN' in col or 'MAX' in col or '_step' in col: continue
            if exclude and exclude.lower() in col_lower: continue
            cols.append(col)

        if not cols:
            print(f'[WARN] No matching columns for keyword "{keyword}" in {file_path}')
            continue

        x = df['total_steps']
        y_mean = df[cols].mean(axis=1).rolling(window=window_size, min_periods=1).mean()
        y_std  = df[cols].std(axis=1).rolling(window=window_size, min_periods=1).mean()

        # Plot line with thicker width and more visible style
        ax.plot(x, y_mean, label=label, color=color, linestyle=lsty, linewidth=lw)
        # Use very light shade to avoid overlap issues
        ax.fill_between(x, y_mean - y_std, y_mean + y_std, color=color, alpha=alpha)

    ax.set_xlabel('Total Steps')
    ax.set_ylabel('Success Rate')
    ax.set_ylim(ylim)
    ax.set_xlim((0,3e6))
    ax.legend(loc='upper left', framealpha=0.9)  # Add framealpha for better legend visibility
    ax.grid(True, linestyle='--', alpha=0.5)
    plt.tight_layout()
    plt.show()
    plt.close(fig)


def plot_causal_effect(plot_config, window_size=50, figsize=(10,6)):
    """
    plot_config: list of dicts, each representing ONE causal-effect CSV, with:
      - path       : str, CSV file path (delta_stag / delta_hare / delta_no type)
      - label      : str, legend label prefix
      - color      : str, matplotlib color
      - linestyle  : str, optional (default '-')
      - linewidth  : float, optional (default 2)
      - keyword    : str, column keyword to match (e.g. 'PCI_rippo')
      - case       : str, one of 'stag' | 'hare' | 'no_reward'
                     determines which column pattern to search for
    """
    plt.close('all')
    plt.rc('font', size=14)
    plt.rc('axes', labelsize=14)
    plt.rc('xtick', labelsize=14)
    plt.rc('ytick', labelsize=14)
    plt.rc('legend', fontsize=14)

    case_patterns = {
        'stag':      ('delta_stag_reward',),
        'hare':      ('delta_forage',),
        'no_reward': ('delta_no_reward',),
    }
    case_linestyles = {
        'stag':      '-',
        'hare':      '-',
        'no_reward': '-',
    }

    fig, ax = plt.subplots(figsize=figsize)
    ax.axhline(y=5, color='gray', linestyle=':', alpha=0.5)
    ax.axhline(y=3, color='gray', linestyle=':', alpha=0.5)

    for entry in plot_config:
        file_path = entry['path']
        label     = entry['label']
        color     = entry['color']
        lw        = entry.get('linewidth', 2)
        keyword   = entry['keyword']
        case      = entry.get('case', 'stag').lower()
        alpha     = entry.get('alpha', 0.15)
        
        patterns = case_patterns.get(case, ('delta_stag_reward',))
        final_lsty = entry.get('linestyle', case_linestyles.get(case, '-'))

        try:
            df = pd.read_csv(file_path)
        except FileNotFoundError:
            print(f'[WARN] File not found: {file_path}')
            continue

        clean_kw = keyword.lower().replace('_', '').replace('-', '')
        cols = []
        for col in df.columns:
            col_lower = col.lower()
            clean_col = col_lower.replace('_', '').replace('-', '')
            if clean_kw not in clean_col: continue
            if not any(p in col_lower for p in patterns): continue
            if 'MIN' in col or 'MAX' in col: continue
            cols.append(col)

        if not cols:
            print(f'[WARN] No matching columns for keyword "{keyword}" / case "{case}" in {file_path}')
            continue

        x = df['total_steps']
        y_mean = df[cols].mean(axis=1).rolling(window=window_size, min_periods=1).mean()
        y_std  = df[cols].std(axis=1).rolling(window=window_size, min_periods=1).mean()

        ax.plot(x, y_mean, label=label, color=color, linestyle=final_lsty, linewidth=lw)
        ax.fill_between(x, y_mean - y_std, y_mean + y_std, color=color, alpha=alpha)

    ax.set_xlabel('Total Steps')
    ax.set_xlim((0,3e6))
    ax.set_ylabel('Causal Shaped Reward')
    ax.legend(loc='upper left')#, bbox_to_anchor=(0.5, 1.0), ncol=3, frameon=False)
    ax.grid(True, linestyle='--', alpha=0.5)
    plt.tight_layout()
    plt.show()
    plt.close(fig)
```

```python
# ── Final Performance Table (RL Paper Style) ─────────────────────────────
import numpy as np

def extract_final_performance(plot_config, window_size=50, last_n=100):
    """
    Extract final performance metrics for each algorithm in plot_config.
    Returns a list of dicts with algorithm performance data.
    """
    results = []

    for entry in plot_config:
        file_path = entry['path']
        label     = entry['label']
        keyword   = entry.get('keyword', 'success_rate')
        exclude   = entry.get('exclude', None)

        try:
            df = pd.read_csv(file_path)
        except FileNotFoundError:
            print(f'[WARN] File not found: {file_path}')
            continue

        # column matching (same as plot function)
        clean_kw = keyword.lower().replace('_', '').replace('-', '')
        cols = []
        for col in df.columns:
            col_lower = col.lower()
            if 'success_rate' not in col_lower: continue
            clean_col = col_lower.replace('_', '').replace('-', '')
            if clean_kw not in clean_col: continue
            if 'MIN' in col or 'MAX' in col or '_step' in col: continue
            if exclude and exclude.lower() in col_lower: continue
            cols.append(col)

        if not cols:
            print(f'[WARN] No matching columns for keyword "{keyword}" in {file_path}')
            continue

        # Extract final performance values
        final_vals = []
        for col in cols:
            series = df[col].dropna()
            smoothed = series.rolling(window=window_size, min_periods=1).mean()
            # Use last_n points average as final performance
            val = smoothed.tail(last_n).mean() if last_n and last_n > 0 else smoothed.iloc[-1]
            final_vals.append(float(val))

        final_vals = np.array(final_vals)
        n_seeds = len(final_vals)

        results.append({
            'algorithm': label,
            'mean': float(np.mean(final_vals)),
            'std': float(np.std(final_vals)),
            'n_seeds': n_seeds,
            'min': float(np.min(final_vals)),
            'max': float(np.max(final_vals)),
        })

    return results

def print_performance_table(plot_config, window_size=50, last_n=200):
    """
    Print a performance table similar to RL papers, showing final success rates.
    """
    results = extract_final_performance(plot_config, window_size=window_size, last_n=last_n)

    if not results:
        print('[ERROR] No performance data extracted.')
        return

    print('\n' + '='*80)
    print('Final Performance Summary (Success Rate)')
    print(f'Based on last {last_n} points with window_size={window_size}')
    print('='*80)

    # Header
    header = f'{"Algorithm":<15} {"N":<5} {"Mean ± Std":<15} {"Min":<8} {"Max":<8}'
    print(header)
    print('-'*80)

    # Results
    for result in results:
        mean_std = f"{result['mean']:.3f}±{result['std']:.3f}"
        print(f"{result['algorithm']:<15} {result['n_seeds']:<5} {mean_std:<15} {result['min']:<8.3f} {result['max']:<8.3f}")

    print('='*80)
    print('Note: Values are success rates (0-1 scale)')
    print('='*80)

# Generate the performance table
print_performance_table(plot_config, window_size=10, last_n=200)
```
